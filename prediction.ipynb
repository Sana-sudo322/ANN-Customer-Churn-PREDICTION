```python
import pickle
import numpy as np
import pandas as pd
from tensorflow.keras.models import load_model
```

```python
#load the model
model = load_model('model.h5')
```

```python
# Load encoders
with open("label_encoder_gender.pkl", "rb") as file:
    label_encoder_gender = pickle.load(file)
```

```python
with open("onehot_encoder_geo.pkl", "rb") as file:
    onehot_encoder_geo = pickle.load(file)
```

```python
# Load scaler
with open("scaler.pkl", "rb") as file:
    scaler = pickle.load(file)
```

```python
customer = {
    "CreditScore": 650,
    "Gender": "Female",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000,
    "Geography": "France"
}
```

```python
gender_encoded = label_encoder_gender.transform(
    [customer["Gender"]]
)[0]

gender_encoded
```

```python
geo_encoded = onehot_encoder_geo.transform(
    [[customer["Geography"]]]
).toarray()

geo_encoded
```

```python
input_data = pd.DataFrame(
    [[
        customer["CreditScore"],
        gender_encoded,
        customer["Age"],
        customer["Tenure"],
        customer["Balance"],
        customer["NumOfProducts"],
        customer["HasCrCard"],
        customer["IsActiveMember"],
        customer["EstimatedSalary"],
        geo_encoded[0][0],
        geo_encoded[0][1],
        geo_encoded[0][2]
    ]],
    columns=[
        "CreditScore",
        "Gender",
        "Age",
        "Tenure",
        "Balance",
        "NumOfProducts",
        "HasCrCard",
        "IsActiveMember",
        "EstimatedSalary",
        "Geography_France",
        "Geography_Germany",
        "Geography_Spain"
    ]
)

input_data
```

```python
input_scaled = scaler.transform(input_data)
```

```python
input_scaled
```

```python
#make prediction
prediction = model.predict(input_scaled)
```

```python
probability = prediction[0][0]

if probability >= 0.5:
    result = "Customer is likely to churn"
else:
    result = "Customer is likely to stay"

print(f"Churn Probability: {probability:.2%}")
print(f"Prediction: {result}")
```
